```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Asset": ["Stocks", "Bonds", "Gold", "Cash"],
    "Weight": [0.50, 0.25, 0.15, 0.10],
    "Expected_Return": [0.12, 0.06, 0.08, 0.03],
    "Risk": [0.25, 0.08, 0.15, 0.01]
}

df = pd.DataFrame(data)

print(df)
```

```python
portfolio_return = np.sum(
    df["Weight"] * df["Expected_Return"]
)

print("Expected Portfolio Return:",
      round(portfolio_return * 100, 2), "%")
```

```python
portfolio_risk = np.sum(
    df["Weight"] * df["Risk"]
)

print("Portfolio Risk Score:",
      round(portfolio_risk * 100, 2), "%")
```

```python
if portfolio_risk < 0.10:
    risk_level = "Low Risk"
elif portfolio_risk < 0.18:
    risk_level = "Medium Risk"
else:
    risk_level = "High Risk"

print("Portfolio Risk Classification:", risk_level)
```

```python
print("===================================")
print("      PORTFOLIO RISK CLASSIFIER")
print("===================================")

print("Expected Return :", round(portfolio_return * 100, 2), "%")
print("Risk Score      :", round(portfolio_risk * 100, 2), "%")
print("Risk Level      :", risk_level)

print("===================================")
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Asset"], df["Weight"] * 100)

plt.title("Portfolio Allocation")
plt.xlabel("Assets")
plt.ylabel("Percentage (%)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Asset"], df["Risk"] * 100)

plt.title("Risk of Each Asset")
plt.xlabel("Assets")
plt.ylabel("Risk (%)")

plt.show()
```
